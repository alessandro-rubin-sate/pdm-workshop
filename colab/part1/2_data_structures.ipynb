## Idioms

### Idiom 1 - Six containers cover almost everything.

`idioms/01_containers.py`

list, tuple, dict, set, NumPy array, DataFrame: what each one is for.

````python
import numpy as np
import polars as pl
````

````python
# list: an ordered collection that can grow   (MATLAB: cell array)
machines = ["M01", "M03"]
machines.append("M08")
print("list: ", machines, "| length", len(machines))
````

````python
# tuple: a small fixed group   (a shape, a pair of results)
shape = (3, 512)
print("tuple:", shape, "| first", shape[0])
````

````python
# dict: values looked up by name   (MATLAB: struct, containers.Map)
settings = {"rpm": 1500, "width": 512}
print("dict: ", settings["rpm"])
````

````python
# set: uniqueness and membership   (MATLAB: unique, ismember)
seen = set(["M01", "M03", "M01"])
print("set:  ", sorted(seen), "| 'M03' in seen:", "M03" in seen)
````

````python
# NumPy array: numbers of one type, used in bulk   (MATLAB: matrix)
levels = np.array([1.0, 2.0, 3.0])
print("array:", levels * 2)
````

````python
# DataFrame: named columns of different types   (MATLAB: table)
table = pl.DataFrame({"machine": machines, "hours": [466.0, 272.0, 582.0]})
print(table)
````

````python
# A list is not an array: predict these before running
print("[1, 2] * 2:  ", [1, 2] * 2)
print("[1, 2] + [3]:", [1, 2] + [3])
print("np.array([1, 2]) * 2:", np.array([1, 2]) * 2)
````

### Idiom 2 - Unpacking: several names in one line.

`idioms/02_unpacking.py`

```
MATLAB:  [rows, cols] = size(A)
```

````python
import numpy as np
````

````python
# One name per element
A = np.zeros((2, 3))
rows, cols = A.shape
print("rows, cols:", rows, cols)
````

````python
# Swap without a temporary variable
low, high = 9, 1
low, high = high, low
print("low, high:", low, high)
````

````python
# A star collects whatever is left
first, *rest = [10, 20, 30]
print("first:", first, "| rest:", rest)
````

````python
# An underscore is a name for a result you do not need   (MATLAB: ~)
quotient, _ = divmod(17, 5)
print("quotient:", quotient)
````

````python
# A function returns one object; several results travel as a tuple
def min_max(x):
    return x.min(), x.max()


signal = np.array([3.0, -1.0, 7.0])
lowest, highest = min_max(signal)
print("lowest, highest:", lowest, highest)
print("without unpacking:", min_max(signal))
````

````python
# The counts must match
try:
    a, b = [1, 2, 3]
except ValueError as error:
    print("ValueError:", error)
````

### Idiom 3 - Loops go over things, not over indices.

`idioms/03_loops.py`

```
MATLAB:  for i = 1:numel(machines), m = machines{i}; ... end
```

````python
machines = ["M01", "M03", "M08"]
lives = [466, 272, 582]
settings = {"rpm": 1500, "width": 512}
````

````python
# The loop variable is the element itself
for machine in machines:
    print(machine)
````

````python
# enumerate: the index as well, when you need it
for i, machine in enumerate(machines):
    print(i, machine)
````

````python
# zip: several sequences, in step
# strict=True raises an error if the lengths differ; without it, zip stops
# silently at the shortest one.
for machine, hours in zip(machines, lives, strict=True):
    print(machine, hours)
````

````python
# A dictionary: keys and values together
for key, value in settings.items():
    print(key, value)
````

````python
# range remains for when you really want the numbers
for i in range(3):
    print("pass", i)
````

### Idiom 4 - Comprehensions: a loop that builds a collection.

`idioms/04_comprehensions.py`

Read them from the middle: `for x in values` first, then what is kept.

````python
values = [1, 2, 3, 4]
machines = ["M01", "M03", "M08"]
lives = [466, 272, 582]
temperature = {"M01": 72.4, "M03": 74.7, "M08": 65.0}
````

````python
# A list
squares = [x**2 for x in values]
print("squares:", squares)
````

````python
# With a filter
hot = [m for m in machines if temperature[m] > 70]
print("hot:", hot)
````

````python
# A dictionary
life = {m: hours for m, hours in zip(machines, lives, strict=True)}
print("life:", life)
````

````python
# Without brackets, straight into a function: no list is built
total = sum(x**2 for x in values)
print("total:", total)
````

````python
# The first one, written out as a loop
squares = []
for x in values:
    squares.append(x**2)
print("squares, the long way:", squares)
````

### Idiom 5 - Dictionaries: values looked up by name.

`idioms/05_dictionaries.py`

```
MATLAB:  a struct with dynamic fields, or containers.Map
```

````python
config = {"machine": "M07", "hours": 100}
````

````python
# Reading
print('config["hours"]:', config["hours"])
````

````python
# A missing key is an error...
try:
    print(config["width"])
except KeyError as error:
    print("KeyError:", error)
````

````python
# ...unless you ask with a default
print('config.get("width", 512):', config.get("width", 512))
print('"hours" in config:', "hours" in config)
````

````python
# Adding or replacing
config["width"] = 256
print(config)
````

````python
# Looping
for key, value in config.items():
    print(key, "->", value)
````

````python
# Nested dictionaries and lists are how settings and JSON files look
experiment = {"machine": "M07", "bands_hz": [10, 50], "model": {"depth": 3}}
print(experiment["model"]["depth"], experiment["bands_hz"][-1])
````

### Trap 1 - Assignment gives a name; it never copies.

`traps/01_names.py`

Predict what each print shows before you run the cell.

````python
# Two names, one list
a = [1, 2, 3]
b = a
b.append(4)
print("a:", a)
print("a is b:", a is b)
````

````python
# The habit: when you want a copy, say so
c = a.copy()
c.append(5)
print("a:", a)
print("c:", c)
print("a is c:", a is c)
````

### Trap 2 - A function gets your object, not a copy.

`traps/02_arguments.py`

The two functions look equivalent. Predict the value of `signal` after each call.

````python
import numpy as np


def remove_offset_in_place(x):
    x -= x.mean()
    return x


def remove_offset(x):
    x = x - x.mean()
    return x
````

````python
# In place: the caller's array changes
signal = np.array([1.0, 2.0, 3.0])
result = remove_offset_in_place(signal)
print("result:", result)
print("signal:", signal)
````

````python
# New object: the caller's array is untouched
signal = np.array([1.0, 2.0, 3.0])
result = remove_offset(signal)
print("result:", result)
print("signal:", signal)
````

### Trap 4 - A copied container shares its contents.

`traps/04_nested_copies.py`

Predict both prints before running.

````python
import copy
````

````python
# One row, referred to twice
grid = [[0] * 3] * 2
grid[0][0] = 1
print("grid:", grid)
````

````python
# A shallow copy: new outer dictionary, same inner list
defaults = {"bands_hz": [10, 50]}
experiment = defaults.copy()
experiment["bands_hz"].append(200)
print("defaults:", defaults)
````

````python
# The habits
grid = [[0] * 3 for _ in range(2)]
grid[0][0] = 1
print("grid built in a loop:", grid)

defaults = {"bands_hz": [10, 50]}
experiment = copy.deepcopy(defaults)
experiment["bands_hz"].append(200)
print("defaults after a deep copy:", defaults)
````

### Trap 5 - Two classics: a default that remembers, a method that returns nothing.

`traps/05_defaults_and_none.py`

Predict every print before running.

````python
# The default list is created once, when the function is defined
def log(value, history=[]):
    history.append(value)
    return history


print(log(1))
print(log(2))
````

````python
# The habit: use None as the default and create the list inside
def log_fixed(value, history=None):
    if history is None:
        history = []
    history.append(value)
    return history


print(log_fixed(1))
print(log_fixed(2))
````

````python
# Methods that modify in place return None
values = [3, 1, 2]
values = values.sort()
print("values:", values)
````

````python
# The habits: sort in place on its own line, or ask for a new list
values = [3, 1, 2]
values.sort()
print("sorted in place:", values)
print("new sorted list:", sorted([3, 1, 2]))
````

### Trap 6 - Indices start at 0, and the end of a slice is excluded.

`traps/06_indexing.py`

```
MATLAB:  x = 10:19;  x(1)  x(end)  x(1:3)
```

````python
import numpy as np
````

````python
# First, last, first three
x = np.arange(10, 20)
print("x[0]:  ", x[0])
print("x[-1]: ", x[-1])
print("x[0:3]:", x[0:3])
````

````python
# Why the end is excluded: lengths and splits need no +1 or -1
a, b, k = 2, 7, 4
print("len(x[a:b]) == b - a:", len(x[a:b]) == b - a)
print("x[:k] then x[k:] rebuilds x:", np.array_equal(np.concatenate([x[:k], x[k:]]), x))
````

````python
# Consecutive windows: where one ends, the next begins
width = 4
for start in range(0, 8, width):
    print(f"x[{start}:{start + width}] ->", x[start : start + width])
````

````python
# range follows the same rule
print("list(range(3)):", list(range(3)))
````

### Trap 7 - A wrong index does not always fail.

`traps/07_silent_index.py`

Which of these lines raise an error? Decide before running.

````python
import numpy as np

x = np.arange(10)
````

````python
# A negative index counts from the end
print("x[-1]:  ", x[-1])
````

````python
# A slice past the end is cut short, without any error
print("x[8:20]:", x[8:20])
````

````python
# Only a single index past the end fails
try:
    print(x[20])
except IndexError as error:
    print("x[20]:   IndexError:", error)
````

````python
# What it costs: an off-by-one that reads the wrong sample
i = 0
print("difference from the previous sample, at i = 0:", x[i] - x[i - 1])
````

````python
# What it costs: a last window shorter than the others
width = 4
for start in range(0, len(x), width):
    window = x[start : start + width]
    print(f"window at {start}: {len(window)} samples")
````
